```python
marks = float(input("Enter your marks: "))

if marks >= 75:
    print("Distinction")
elif marks >= 60:
    print("First Class")
elif marks >= 50:
    print("Second Class")
elif marks >= 35:
    print("Pass")
else:
    print("Fail")
```
