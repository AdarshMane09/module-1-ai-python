```python
name = input("Enter student name: ")

english = float(input("Enter English marks: "))
python = float(input("Enter Python marks: "))
dbms = float(input("Enter DBMS marks: "))

total = english + python + dbms
percentage = (total / 300)*100

print("\n----- Student Result -----")
print("Name:", name)
print("English:", english)
print("Python:", python)
print("DBMS:", dbms)
print("Total:", total)
print("Percentage:", percentage)

if percentage >= 75:
    print("Grade: A")
elif percentage >= 60:
    print("Grade: B")
elif percentage >= 50:
    print("Grade: C")
elif percentage >= 35:
    print("Grade: D")
else:
    print("Grade: F")
```
